```python
# 00_distress_radar_workflow.ipynb
# From-scratch pipeline: advanced search -> features -> signals -> scores & rankings

# ==============================
# 0. IMPORTS & BASIC SETUP
# ==============================
import os
import json
from pathlib import Path
from datetime import datetime, date

import requests
import pandas as pd
import matplotlib.pyplot as plt
from dotenv import load_dotenv
from IPython.display import display

plt.rcParams["figure.figsize"] = (8, 5)
plt.rcParams["axes.grid"] = True
plt.rcParams["figure.dpi"] = 110

# --- Project root detection ---
here = Path().resolve()
if (here / "data").exists():
    project_root = here
elif (here.parent / "data").exists():
    project_root = here.parent
else:
    raise RuntimeError("Could not find data/ directory from this notebook.")

print("project_root:", project_root)

data_dir      = project_root / "data"
raw_dir       = data_dir / "raw"
profiles_dir  = raw_dir / "profiles"
filings_dir   = raw_dir / "filings"
charges_dir   = raw_dir / "charges"
processed_dir = data_dir / "processed"

profiles_dir.mkdir(parents=True, exist_ok=True)
filings_dir.mkdir(parents=True, exist_ok=True)
charges_dir.mkdir(parents=True, exist_ok=True)
processed_dir.mkdir(parents=True, exist_ok=True)

# --- API auth ---
load_dotenv(project_root / ".env")
API_KEY = os.getenv("CH_API_KEY")
if not API_KEY:
    raise RuntimeError("CH_API_KEY not found in .env")
print("API key loaded OK.")

BASE_URL = "https://api.company-information.service.gov.uk"
TODAY    = date.today()

session = requests.Session()
session.auth = (API_KEY, "")          # API key as username, blank password
session.headers.update({"Accept": "application/json"})


# ==============================
# 0A. USER–TUNABLE CONFIG
# ==============================

# --- Seed universe / sample size ---
SEED_SIC_CODES      = [70229, 70221, 73110]   # B2B services cluster
SEED_COMPANY_STATUS = "active"
SEED_SIZE           = 600                     # how many results from advanced search (max 5000)
SEED_LOCATION       = None                    # e.g. "london" or None

# How many companies to build features for (None = all in seed)
MAX_COMPANIES_FEATURES = None   # e.g. 200, 500, None

# Items per page for filings / charges endpoints
FILINGS_ITEMS_PER_PAGE  = 200
CHARGES_ITEMS_PER_PAGE  = 200

# Lookback window for "num_accounts_last_10y"
ACCOUNTS_LOOKBACK_YEARS = 10

# --- Distress signal assumptions ---
DISTRESS_TRIGGER_OVERDUE_ACCOUNTS     = True   # any "overdue" flag is serious
DISTRESS_MIN_LATE_ACCOUNTS            = 2      # need at least 2 late filings
DISTRESS_MIN_OUTSTANDING_CHARGES      = 3      # 3+ live charges = non-trivial leverage
DISTRESS_YOUNG_AGE_YEARS              = 5.0    # "young" company threshold
DISTRESS_YOUNG_MIN_TOTAL_CHARGES      = 2      # young but already has 2+ charges
DISTRESS_SCORE_OUTSTANDING_THRESHOLD  = 3      # +1 point if outstanding >= this

# --- Boring-but-profitable assumptions ---
BORING_MIN_AGE_YEARS                  = 10.0   # at least 10 years trading
BORING_MIN_ACCOUNTS_LAST_10Y          = 7      # ~70% of years with accounts
BORING_MAX_LATE_ACCOUNTS              = 0      # no late filings
BORING_REQUIRE_NOT_OVERDUE            = True   # currently not overdue
BORING_MAX_OUTSTANDING_CHARGES        = 2      # allow some debt, not a lot
BORING_SCORE_AGE_THRESHOLD            = 12.0   # +1 if 12+ years old
BORING_SCORE_ACCOUNTS_THRESHOLD       = 8      # +1 if 8+ accounts in last 10y

# --- Score bucketing & rankings ---
SCORE_BUCKET_BINS    = (0, 34, 67, 101)
SCORE_BUCKET_LABELS  = ("Low", "Medium", "High")
TOP_N_DISTRESSED      = 20
TOP_N_BORING          = 20

# --- Visualisation parameters ---
HIST_BINS             = 10


# ==============================
# 1. API HELPERS
# ==============================
def api_get(endpoint: str, params: dict | None = None) -> dict:
    url = BASE_URL + endpoint
    resp = session.get(url, params=params, timeout=10)
    resp.raise_for_status()
    return resp.json()


def fetch_company_profile(company_number: str) -> dict:
    cn = str(company_number).strip()
    return api_get(f"/company/{cn}")


def fetch_filing_history(company_number: str) -> dict:
    cn = str(company_number).strip()
    params = {"items_per_page": FILINGS_ITEMS_PER_PAGE, "start_index": 0}
    return api_get(f"/company/{cn}/filing-history", params=params)


def fetch_charges(company_number: str) -> dict:
    cn = str(company_number).strip()
    params = {"items_per_page": CHARGES_ITEMS_PER_PAGE, "start_index": 0}
    return api_get(f"/company/{cn}/charges", params=params)


def save_json(path: Path, obj: dict) -> None:
    """Optional: store raw API responses for later inspection."""
    path.parent.mkdir(parents=True, exist_ok=True)
    with path.open("w", encoding="utf-8") as f:
        json.dump(obj, f, ensure_ascii=False, indent=2)


# ==============================
# 2. SEED UNIVERSE (ADVANCED SEARCH)
# ==============================
def build_seed_universe(
    sic_codes,
    company_status: str,
    size: int,
    location: str | None = None,
) -> pd.DataFrame:
    """Always build seed via /advanced-search (no reuse, no checks)."""
    if isinstance(sic_codes, (list, tuple, set)):
        sic_param = ",".join(str(code) for code in sic_codes)
    else:
        sic_param = str(sic_codes)

    params = {
        "sic_codes": sic_param,
        "company_status": company_status,
        "size": size,
        "start_index": 0,
    }
    if location:
        params["location"] = location

    result = api_get("/advanced-search/companies", params=params)
    items = result.get("items", []) or []

    rows = []
    for item in items:
        roa = item.get("registered_office_address") or {}
        address_parts = [
            roa.get("address_line_1"),
            roa.get("address_line_2"),
            roa.get("locality"),
            roa.get("region"),
            roa.get("postal_code"),
            roa.get("country"),
        ]
        address = ", ".join(part for part in address_parts if part)

        rows.append(
            {
                "company_number": item.get("company_number"),
                "name": item.get("company_name"),
                "sic_codes": ";".join(item.get("sic_codes") or []),
                "registered_office_address": address,
            }
        )

    seed_df = pd.DataFrame(rows)
    seed_csv = data_dir / "companies_seed.csv"
    seed_csv.parent.mkdir(parents=True, exist_ok=True)
    seed_df.to_csv(seed_csv, index=False)
    print(f"[SEED] Built and saved seed universe ({len(seed_df)} companies) to {seed_csv}")
    return seed_df


seed_df = build_seed_universe(
    sic_codes=SEED_SIC_CODES,
    company_status=SEED_COMPANY_STATUS,
    size=SEED_SIZE,
    location=SEED_LOCATION,
)
print("Seed universe shape:", seed_df.shape)
display(seed_df.head())


# ==============================
# 3. FEATURE EXTRACTION (DIRECT FROM API)
# ==============================
FEATURE_COLUMNS = [
    # Basic
    "company_number",
    "name",
    "incorporation_date",
    "status",
    "sic_codes",
    "age_years",
    # Filing behaviour
    "num_accounts_last_10y",
    "num_late_accounts",
    "last_accounts_date",
    "overdue_accounts_flag",
    # Charges
    "num_charges_total",
    "num_charges_outstanding",
    "num_charges_satisfied",
    "has_floating_charge_flag",
    "last_charge_date",
]


def parse_date(date_str: str | None):
    if not date_str:
        return None
    try:
        return datetime.fromisoformat(date_str).date()
    except ValueError:
        return None


def extract_basic_features(profile: dict) -> dict:
    incorporation_date = parse_date(profile.get("date_of_creation"))
    sic_codes = profile.get("sic_codes") or []
    if isinstance(sic_codes, (list, tuple, set)):
        sic_str = ";".join(str(code) for code in sic_codes)
    else:
        sic_str = str(sic_codes) if sic_codes else ""

    age_years = None
    if incorporation_date:
        age_years = round((TODAY - incorporation_date).days / 365.25, 2)

    return {
        "company_number": profile.get("company_number"),
        "name": profile.get("company_name"),
        "incorporation_date": incorporation_date,
        "status": profile.get("company_status"),
        "sic_codes": sic_str,
        "age_years": age_years,
    }


def extract_filing_features(profile: dict, filing_history: dict) -> dict:
    items = filing_history.get("items", []) or []
    ten_years_ago = date(TODAY.year - ACCOUNTS_LOOKBACK_YEARS, TODAY.month, TODAY.day)

    account_dates = []
    num_accounts_last_10y = 0
    num_late_accounts = 0

    for item in items:
        if item.get("category") != "accounts":
            continue
        d = parse_date(item.get("date"))
        if not d:
            continue
        if d >= ten_years_ago:
            num_accounts_last_10y += 1
            account_dates.append(d)
            desc = (item.get("description") or "").lower()
            if "late" in desc or item.get("is_late"):
                num_late_accounts += 1

    last_accounts_date = max(account_dates) if account_dates else None

    accounts_info = profile.get("accounts") or {}
    next_due = parse_date(accounts_info.get("next_due"))
    overdue_accounts_flag = bool(next_due and next_due < TODAY)

    return {
        "num_accounts_last_10y": num_accounts_last_10y,
        "num_late_accounts": num_late_accounts,
        "last_accounts_date": last_accounts_date,
        "overdue_accounts_flag": overdue_accounts_flag,
    }


def extract_charge_features(charges: dict) -> dict:
    items = charges.get("items", []) or []

    num_charges_total = charges.get("total_count", len(items))
    num_charges_satisfied = 0
    num_charges_outstanding = 0
    has_floating_charge_flag = False
    dates = []

    for item in items:
        satisfied_on = parse_date(item.get("satisfied_on"))
        created_on = parse_date(item.get("created_on") or item.get("delivered_on"))

        if satisfied_on:
            num_charges_satisfied += 1
            dates.append(satisfied_on)
        else:
            num_charges_outstanding += 1

        if created_on:
            dates.append(created_on)

        classification = item.get("classification") or {}
        if "floating" in (classification.get("type") or "").lower():
            has_floating_charge_flag = True
        if item.get("floating_charge"):
            has_floating_charge_flag = True

    last_charge_date = max(dates) if dates else None

    return {
        "num_charges_total": num_charges_total,
        "num_charges_outstanding": num_charges_outstanding,
        "num_charges_satisfied": num_charges_satisfied,
        "has_floating_charge_flag": has_floating_charge_flag,
        "last_charge_date": last_charge_date,
    }


def build_company_feature_row(company_number: str) -> dict:
    """Call the 3 endpoints for a company, save raw JSON (optional), and compute features."""
    cn = str(company_number).strip()

    profile = fetch_company_profile(cn)
    filings = fetch_filing_history(cn)
    charges = fetch_charges(cn)

    # Optional: save raw JSON for inspection
    save_json(profiles_dir / f"{cn}.json", profile)
    save_json(filings_dir  / f"{cn}.json", filings)
    save_json(charges_dir  / f"{cn}.json", charges)

    feat = {}
    feat.update(extract_basic_features(profile))
    feat.update(extract_filing_features(profile, filings))
    feat.update(extract_charge_features(charges))
    return feat


def build_features_for_seed(seed_df: pd.DataFrame) -> pd.DataFrame:
    records = []

    it = seed_df
    if MAX_COMPANIES_FEATURES is not None:
        it = it.head(MAX_COMPANIES_FEATURES)

    total = len(it)
    for i, row in enumerate(it.itertuples(index=False), start=1):
        cn = row.company_number
        try:
            rec = build_company_feature_row(cn)
            records.append(rec)
        except Exception as e:
            print(f"[FEATURES][ERROR] {cn}: {e}")
        if i % 20 == 0 or i == total:
            print(f"[FEATURES] Processed {i}/{total} companies")

    features_df = pd.DataFrame(records, columns=FEATURE_COLUMNS)

    # Fix dtypes
    for col in ["incorporation_date", "last_accounts_date", "last_charge_date"]:
        if col in features_df.columns:
            features_df[col] = pd.to_datetime(features_df[col], errors="coerce")

    for col in ["overdue_accounts_flag", "has_floating_charge_flag"]:
        if col in features_df.columns:
            features_df[col] = features_df[col].fillna(False).astype(bool)

    out_path = data_dir / "company_features_week2.csv"
    features_df.to_csv(out_path, index=False)
    print("Saved feature table to:", out_path)
    return features_df


features_df = build_features_for_seed(seed_df)

print("\n[FEATURES] Dtypes:")
print(features_df.dtypes)
display(features_df.head())


# ==============================
# 4. SIGNALS & RAW SCORES
# ==============================
def add_signals(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()

    # Safety cleaning
    for col in [
        "num_accounts_last_10y", "num_late_accounts",
        "num_charges_total", "num_charges_outstanding",
        "num_charges_satisfied"
    ]:
        if col in df.columns:
            df[col] = df[col].fillna(0)

    if "age_years" in df.columns:
        df["age_years"] = df["age_years"].fillna(0)

    if "overdue_accounts_flag" in df.columns:
        df["overdue_accounts_flag"] = df["overdue_accounts_flag"].astype(bool)
    if "has_floating_charge_flag" in df.columns:
        df["has_floating_charge_flag"] = df["has_floating_charge_flag"].astype(bool)

    # --- Distress ---
    distress_overdue = df.get("overdue_accounts_flag", False) & DISTRESS_TRIGGER_OVERDUE_ACCOUNTS
    distress_late = df.get("num_late_accounts", 0) >= DISTRESS_MIN_LATE_ACCOUNTS
    distress_many_charges = df.get("num_charges_outstanding", 0) >= DISTRESS_MIN_OUTSTANDING_CHARGES
    distress_young_with_debt = (
        (df.get("age_years", 0) < DISTRESS_YOUNG_AGE_YEARS) &
        (df.get("num_charges_total", 0) >= DISTRESS_YOUNG_MIN_TOTAL_CHARGES)
    )

    df["signal_potential_distress"] = (
        distress_overdue | distress_late | distress_many_charges | distress_young_with_debt
    )

    # --- Boring-but-profitable ---
    boring_old_enough = df.get("age_years", 0) >= BORING_MIN_AGE_YEARS
    boring_enough_accounts = df.get("num_accounts_last_10y", 0) >= BORING_MIN_ACCOUNTS_LAST_10Y
    boring_no_late = df.get("num_late_accounts", 0) <= BORING_MAX_LATE_ACCOUNTS

    if BORING_REQUIRE_NOT_OVERDUE:
        boring_not_overdue = df.get("overdue_accounts_flag", False) == False
    else:
        boring_not_overdue = True

    boring_low_outstanding = df.get("num_charges_outstanding", 0) <= BORING_MAX_OUTSTANDING_CHARGES

    df["signal_boring_profitable_like"] = (
        boring_old_enough &
        boring_enough_accounts &
        boring_no_late &
        boring_not_overdue &
        boring_low_outstanding
    )

    # --- Raw scores (0–3) ---
    df["distress_score_v1"] = (
        df["overdue_accounts_flag"].astype(int)
        + (df["num_late_accounts"] >= DISTRESS_MIN_LATE_ACCOUNTS).astype(int)
        + (df["num_charges_outstanding"] >= DISTRESS_SCORE_OUTSTANDING_THRESHOLD).astype(int)
    )

    df["boring_profitable_score_v1"] = (
        (df["age_years"] >= BORING_SCORE_AGE_THRESHOLD).astype(int)
        + (df["num_accounts_last_10y"] >= BORING_SCORE_ACCOUNTS_THRESHOLD).astype(int)
        + (df["num_late_accounts"] <= BORING_MAX_LATE_ACCOUNTS).astype(int)
    )

    return df


features_with_signals = add_signals(features_df)
features_with_signals_path = processed_dir / "company_features_with_signals_v1.csv"
features_with_signals.to_csv(features_with_signals_path, index=False)
print("\n[FEATURES] Saved with signals to:", features_with_signals_path)

print("\nSignal value counts:")
print("Potential distress:")
print(features_with_signals["signal_potential_distress"].value_counts(dropna=False))
print("\nBoring-profitable-like:")
print(features_with_signals["signal_boring_profitable_like"].value_counts(dropna=False))

print("\n[FEATURES] Preview with signals:")
display(
    features_with_signals[[
        "company_number", "name",
        "age_years",
        "num_accounts_last_10y", "num_late_accounts", "overdue_accounts_flag",
        "num_charges_total", "num_charges_outstanding",
        "signal_potential_distress", "distress_score_v1",
        "signal_boring_profitable_like", "boring_profitable_score_v1"
    ]].head(10)
)


# ==============================
# 5. SCORE SCALING (0–100) & BUCKETS
# ==============================
def scale_min_max(series: pd.Series) -> pd.Series:
    s = series.astype(float)
    if s.empty:
        return s
    feature_min = s.min()
    feature_max = s.max()
    if feature_min == feature_max:
        return pd.Series(50.0, index=s.index)
    scaled = (s - feature_min) / (feature_max - feature_min)
    return (scaled.clip(0, 1) * 100).round(1)


def bucket_score(score_series: pd.Series) -> pd.Series:
    return pd.cut(
        score_series,
        bins=SCORE_BUCKET_BINS,
        labels=SCORE_BUCKET_LABELS,
        right=False,
        include_lowest=True,
    )


def compute_scores(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()
    if "distress_score_v1" not in df.columns or "boring_profitable_score_v1" not in df.columns:
        raise KeyError("Missing raw score columns in features_with_signals.")

    df["distress_score_0_100"] = scale_min_max(df["distress_score_v1"])
    df["distress_bucket"]      = bucket_score(df["distress_score_0_100"])

    df["boring_profitable_score_0_100"] = scale_min_max(df["boring_profitable_score_v1"])
    df["boring_profitable_bucket"]      = bucket_score(df["boring_profitable_score_0_100"])

    out_path = processed_dir / "company_scores_v1.csv"
    df.to_csv(out_path, index=False)
    print("\n[SCORING] Saved company_scores_v1 to:", out_path)
    return df


scores_df = compute_scores(features_with_signals)
print("\n[SCORING] Head:")
display(scores_df.head())


# ==============================
# 6. TOP LISTS / RANKINGS
# ==============================
BASE_COLS = [
    "company_number",
    "name",
    "age_years",
    "last_accounts_date",
    "num_charges_outstanding",
    "status",
]

# Top distressed
distressed_cols = BASE_COLS + [
    "distress_score_0_100",
    "distress_bucket",
    "signal_potential_distress" if "signal_potential_distress" in scores_df.columns else None,
]
top_distressed = (
    scores_df.sort_values("distress_score_0_100", ascending=False)
             .head(TOP_N_DISTRESSED)[distressed_cols]
).dropna(axis=1, how="all")

print(f"\nTop {TOP_N_DISTRESSED} distressed:")
display(top_distressed)

# Top boring-profitable
boring_cols = BASE_COLS + [
    "boring_profitable_score_0_100",
    "boring_profitable_bucket",
    "signal_boring_profitable_like" if "signal_boring_profitable_like" in scores_df.columns else None,
]
top_boring = (
    scores_df.sort_values("boring_profitable_score_0_100", ascending=False)
             .head(TOP_N_BORING)[boring_cols]
).dropna(axis=1, how="all")

print(f"\nTop {TOP_N_BORING} boring-profitable:")
display(top_boring)

# Save ranking views
out_top_distressed = processed_dir / f"top_{TOP_N_DISTRESSED}_distressed_v1.csv"
out_top_boring     = processed_dir / f"top_{TOP_N_BORING}_boring_profitable_v1.csv"

top_distressed.to_csv(out_top_distressed, index=False)
top_boring.to_csv(out_top_boring, index=False)

print("\nSaved ranking views:")
print("  ", out_top_distressed)
print("  ", out_top_boring)

# Full watchlists (flagged only)
distress_watchlist = scores_df[scores_df["signal_potential_distress"] == True]\
    .sort_values(["distress_score_0_100", "age_years"], ascending=[False, True])
boring_watchlist = scores_df[scores_df["signal_boring_profitable_like"] == True]\
    .sort_values(["boring_profitable_score_0_100", "age_years"], ascending=[False, False])

distress_watchlist.to_csv(processed_dir / "distress_watchlist_v1.csv", index=False)
boring_watchlist.to_csv(processed_dir / "boring_profitable_watchlist_v1.csv", index=False)

# ==============================
# 7. BASIC VISUALISATIONS
# ==============================

print("\nDistress score summary:")
print(scores_df["distress_score_0_100"].describe())
print("\nBoring-profitable score summary:")
print(scores_df["boring_profitable_score_0_100"].describe())

print("\nDistress buckets:")
print(scores_df["distress_bucket"].value_counts(dropna=False))
print("\nBoring-profitable buckets:")
print(scores_df["boring_profitable_bucket"].value_counts(dropna=False))

# 7.1 Histogram: distress scores
plt.figure()
scores_df["distress_score_0_100"].dropna().hist(bins=HIST_BINS)
plt.xlabel("Distress score (0–100)")
plt.ylabel("Number of companies")
plt.title("Distribution of distress scores")
plt.tight_layout()
plt.show()

# 7.2 Histogram: boring-profitable scores
plt.figure()
scores_df["boring_profitable_score_0_100"].dropna().hist(bins=HIST_BINS)
plt.xlabel("Boring-profitable score (0–100)")
plt.ylabel("Number of companies")
plt.title("Distribution of boring-profitable scores")
plt.tight_layout()
plt.show()

# 7.3 Scatter: distress vs age
plt.figure()
tmp = scores_df.dropna(subset=["age_years", "distress_score_0_100"])
plt.scatter(tmp["age_years"], tmp["distress_score_0_100"], alpha=0.7)
plt.xlabel("Company age (years)")
plt.ylabel("Distress score (0–100)")
plt.title("Distress score vs company age")
plt.tight_layout()
plt.show()

# 7.4 Scatter: boring-profitable vs age
plt.figure()
tmp = scores_df.dropna(subset=["age_years", "boring_profitable_score_0_100"])
plt.scatter(tmp["age_years"], tmp["boring_profitable_score_0_100"], alpha=0.7)
plt.xlabel("Company age (years)")
plt.ylabel("Boring-profitable score (0–100)")
plt.title("Boring-profitable score vs company age")
plt.tight_layout()
plt.show()
```
